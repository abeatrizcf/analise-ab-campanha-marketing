## *Cálculo do tamanho amostral mínimo*
##### Através da Biblioteca statsmodels calcular o tamanho amostral mínimo teórico das amostras dos grupos A e B e compará-los com o número de amostrar presentes na base do Kaggle


#### Importações necessárias

```python
from statsmodels.stats.proportion import proportion_effectsize
from statsmodels.stats.power import NormalIndPower
```

#### Definição das variáveis

```python
baseline = 0.05
mde = 0.01

alpha = 0.05
poder_estatistico = 0.80

p1 = baseline
p2 = baseline + mde
```

##### Cálculo do effect size, ou seja, o quão grande é a diferença entre os dois grupos, levando em consideração as proporções dos mesmo

```python
effect_size = proportion_effectsize(p1, p2)
```

##### Cálculo do tamanho amostral

```python
sample_size = NormalIndPower().solve_power(
    effect_size=effect_size,
    alpha=alpha,
    power=poder_estatistico,
    ratio=1,
    alternative="two-sided"
)
print(sample_size)
```

#### Conclusão:

##### Conclui-se, portanto, que os grupos (considerando uma proporção 1:1) devem ter no mínimo 8142 amostras
