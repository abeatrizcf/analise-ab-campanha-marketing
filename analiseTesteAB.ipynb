### Cálculo da taxa de conversão de cada grupo

```python
import pandas as pd
from statsmodels.stats.proportion import proportions_ztest  
df = pd.read_csv('marketing_AB.csv', index_col=0)
```

```python
taxa_de_conversao = df.groupby('test group')['converted'].mean()


print(f"Taxa de conversão com propaganda: {taxa_de_conversao['ad']:.4%}")
print(f"Taxa de conversão sem propaganda: {taxa_de_conversao['psa']:.4%}")
```

### Cálculo do número de conversões de cada grupo

```python
conversoes = df.groupby('test group')['converted'].sum()
conversoes_com_propaganda = conversoes['ad']
conversoes_sem_propaganda = conversoes['psa']
conversoes_total = [conversoes_com_propaganda, conversoes_sem_propaganda]
print(f'Conversões com propaganda: {conversoes_com_propaganda}')
print(f'Conversões sem propaganda: {conversoes_sem_propaganda}')
```

### Cálculo do número de registros de cada grupo

```python
registros = df.groupby('test group')['converted'].count()
registros_com_propaganda = registros['ad']
registros_sem_propaganda = registros['psa']
registros_total = [registros_com_propaganda, registros_sem_propaganda]
print(f"Registros com propaganda: {registros_com_propaganda}")
print(f"Registros sem propaganda: {registros_sem_propaganda}")  
```

### Cálculo do Z-stat e do p-valor

```python
z_stat, p_valor = proportions_ztest(conversoes_total, registros_total)
## obs: sem especificar alternative, o teste é bicaudal por padrão
print(f"Estatística Z: {z_stat:.4f}")
print(f"P-valor: {p_valor:.2e}")
```

### Análise da conversão orgânica e ganho incremental

```python
diferenca_taxa = taxa_de_conversao['ad'] - taxa_de_conversao['psa']
print(f"Diferença de taxas: {diferenca_taxa:.4%}")
```

### Cálculo do ganho incremental estimado
##### >> Ou seja, aquele que estima-se como oriundo da utilização das propagandas

```python
ganho_incremental  =registros_com_propaganda * diferenca_taxa
print(f"Ganho incremental: {ganho_incremental:.0f}")
```

### Cálculo da conversão orgânica estimada
##### >> Ou seja, aquela esperada independente do uso de anúncios

```python
conversao_organica = registros_com_propaganda * taxa_de_conversao['psa']
print(f"Conversão orgânica: {conversao_organica:.0f}")  
```

#### Conferência dos valores estimados retornando para o número total de conversão achado anteriormente

```python
print(f"Conversão total: {conversao_organica + ganho_incremental:.0f} ")
```
