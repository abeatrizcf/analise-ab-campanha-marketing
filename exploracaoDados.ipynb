#### EDA dos dados do banco Marketing A/B Testing:

1. Verificação da quantidade, nome e tipo das colunas 
2. Verificação da quantidade de valores não nulos por coluna
2. Verificação de duplicadas
3. Verificação de linhas vazias


```python
import pandas as pd
df = pd.read_csv('marketing_AB.csv', index_col=0)
df.head()
#df.info()
```

##### verificação de linhas duplicadas e da quantidade de amostras de cada grupo

```python
print ( "linhas duplicadas: ", df.duplicated().sum())
print ( "amostras por grupo: ", df['test group'].value_counts())
```

Cálculo da taxa de conversão de cada grupo

```python
taxa_de_conversao = df.groupby('test group')['converted'].mean()


print(f"Taxa de conversão com propaganda: {taxa_de_conversao['ad']:.3%}")
print(f"Taxa de conversão sem propaganda: {taxa_de_conversao['psa']:.3%}")
```
